# test and train data Split

```python
Y = df['class']
X = df.drop('class', axis = 1)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X,Y, test_size = 0.33, random_state = 100)
```

```python
x_train.head()
```

```python
x_test.head()
```

```python
y_train.head()
```

```python
y_test.head()
```

```python
# doing standard scaling to assign proper weighatge for the values
```

```python
# feature scalling
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred[0:10]
```

```python
for i in (y_pred[0:10]):
    if y_pred[i] == 1:
        print('It is spam ')
    else :
        print('it is not a spam')
```

```python
y_pred[0:100]
```

```python
# evaluating the model using model evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# trying other model to verfiy wheather we get better results or not
# in next step we are using a better model like decision tree
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(x_train, y_train)
```

```python
dt_pred = dt_model.predict(x_test)
```

```python
dt_pred[0:50]
```

```python
# evaluating the model using model evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, dt_pred))

print("\nClassification Report:")
print(classification_report(y_test, dt_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, dt_pred))
```
